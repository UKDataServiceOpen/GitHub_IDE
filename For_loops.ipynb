# Coding Basics: For Loops

Working with lists, sets, and dataframes


## Lists

### Example no1. Converting percentages to proportions

```python
# a list we will iterate over with our for loop
percentages = [25, 50, 75, 90]
```

```python
# confirm data type
type(percentages)
```

```python
# print elements in list
print(percentages)
```

```python
# use for loop to convert percentages to  proportions
for pct in percentages:
    print(pct/100)
```

```python
# assign new proportions to a new list

proportions = []

for pct in percentages:
    prop = pct/100
    # use append to populate new list
    proportions.append(prop)
```

```python
print(proportions)
```

## Sets

### Example no2. Counting categorical data

```python
responses = ["UK", "Germany", "France", "Germany"]
```

```python
countries = set(responses)
```

```python
countries
```

```python
for country in countries:
    count = responses.count(country)
    print(country, ":", count)
```

## Dataframes

### Example no3. Converting raw counts into percentages

```python
import pandas as pd
```

```python
df = pd.read_excel("data/test_data.xlsx")
```

```python
df
```

#### Using .iterrows()

```python
for index, row in df.iterrows():
    percentage = (row["n_agreed"] / row["n_total"]) * 100
    print(round(percentage, 2))
```

#### Using range() and len()


```python
for i in range(len(df)):
    percentage = (df["n_agreed"][i] / df["n_total"][i]) * 100
    print(df["region"][i], ":", round(percentage, 1), "%")
```

```python
### Using list comprehension

df["percentages"] = [round((df["n_agreed"][i] / df["n_total"][i]) * 100, 1) for i in range(len(df))]
```

```python
# OR

df["percentages"] = [round((row["n_agreed"] / row["n_total"]) * 100, 1) for _, row in df.iterrows()]
```
